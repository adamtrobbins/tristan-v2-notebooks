```python
# Imports and Folder Selection

from graphet import Data
from graphet.plugins import TristanV2 
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')
import sys
sys.path.append('../v2_proc_files')
from useful_funcs import *
from TristanRun import TristanRun

%matplotlib widget

%load_ext autoreload
%autoreload 2 

set_standard_plot_params()

```

```python
masses = {1: 1.0, 2:25.0}
species = {1: "electron", 2: "ion"}
colors = {1: "blue", 2: "orange"}

plt.close('all')


for i in [1, 2]:
    bin_centers = np.load(f"gammap_bin_centers_{i}.npy")
    hist = np.load(f"gammap_hist_{i}.npy")

    mass = masses[i]

    integral = np.sum(hist * np.diff(bin_centers)[0])
    hist /= integral * mass

    plt.loglog(mass * (bin_centers - 1), hist, label=f"{species[i]}", color=colors[i])

metadata = {"run": "23_test", "notes": "good 3.6k run, final timestep"}

plt.title("Exhaust Spectra in Flow Frame")
plt.xlabel("$(m / m_e) (\gamma' - 1)$")
plt.ylabel("$f(\gamma')$")
plt.legend()

save_polished_fig(f"exhaust_spectra_04_28_26.png", dpi = 300, metadata = metadata)

plt.show()
```
