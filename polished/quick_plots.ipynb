```python
# Plot energy frac

import sys

import numpy as np
import pandas as pd

import matplotlib
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')

sys.path.append('..')
from useful_funcs import *

%matplotlib widget

plt.rcParams['text.usetex'] = True
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.size'] = 12
```

```python
pfrac_dict_filepath = "../fields/pfracs.npy"

pfracs = np.load(pfrac_dict_filepath, allow_pickle=True).item()
```

```python
# Separate the data based on sigma values
sigma_03 = {k[1]: v for k, v in pfracs.items() if k[0] == 0.3}
sigma_10 = {k[1]: v for k, v in pfracs.items() if k[0] == 10}

# Sort the data by ylim
sigma_03 = dict(sorted(sigma_03.items()))
sigma_10 = dict(sorted(sigma_10.items()))

# Format the y-values as percentages
sigma_03 = {k: v * 100 for k, v in sigma_03.items()}
sigma_10 = {k: v * 100 for k, v in sigma_10.items()}

# Plot the data
plt.figure()
plt.plot(sigma_10.keys(), sigma_10.values(), label='$\sigma$ = 10', marker='s', color = '#ff7f0e')
plt.plot(sigma_03.keys(), sigma_03.values(), label='$\sigma$ = 0.3', marker='s', color = '#1f77b4')


# Annotate the data points

annotate = False

#if annotate:
#    for x, y in sigma_03.items():
#        if (x != 6000):
#            plt.annotate(f'({x:.0f}, {y:.0f}\%)', (x, y), textcoords="offset points", xytext=(20,10), ha='center')
#        else:
#            plt.annotate(f'({x:.0f}, {y:.1f}\%)', (x, y), textcoords="offset points", xytext=(-15, 15), ha='center')
#    for x, y in sigma_10.items():
#        plt.annotate(f'({x:.0f}, {y:.0f}\%)', (x, y), textcoords="offset points", xytext=(20,-20), ha='center')

run_dict = {
    (0.3, 240): 'II',
    (0.3, 480): 'IV',
    (0.3, 6000): 'V',
    (10, 240): 'I',
    (10, 480): 'III',
}

def get_label_str(x, y):
    return f'Run {run_dict[(x, y)]}'

for x, y in sigma_03.items():
    if (x != 6000):
        plt.annotate(get_label_str(0.3, x), (x, y), textcoords="offset points", xytext=(20,10), ha='center')
    else:
        plt.annotate(get_label_str(0.3, x), (x, y), textcoords="offset points", xytext=(-15, 15), ha='center')
for x, y in sigma_10.items():
    plt.annotate(get_label_str(10, x), (x, y), textcoords="offset points", xytext=(10,-20), ha='center')

plt.xlabel('$L_y$ [$c / \omega_p$]')
plt.ylabel('Plasmoid Energy Fraction [\%]')
plt.xscale('log')
plt.ylim(0, 100)
plt.legend(fontsize=14)
plt.legend()
plt.show()

# Add more labelled numbers (ticks) on the bottom axis
plt.xticks([240, 480, 6000], ['$240$', '$480$', '$6000$'])

save_polished_fig("pfrac")
```
