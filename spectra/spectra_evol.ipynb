```python
# Purpose - compare spectra over time

findex = 130
species = 1
step_choice = "last"
```

```python
# Imports and Folder Selection

import sys

import numpy as np
import pandas as pd

import matplotlib
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')

from graphet import Data
from graphet.plugins import TristanV2 

sys.path.append('..')
from useful_funcs import *
sys.path.append('../fmgmt')
from TristanRun import TristanRun

plt.close('all')
%matplotlib widget
set_standard_plot_params()

# Read Input File Data

run = TristanRun(findex)
boilerplate = run.boilerplate #run.boilerplate + f"tstep = {run.steps[0]}"

tstep_choices = {"first": [run.steps[0]], "last": [run.steps[-1]], "all": run.steps}

my_steps = tstep_choices[step_choice]
my_steps = [85, 87, 89]
```

```python
d = run.config_data(my_steps, mode = "spectra")

energy_bins = np.insert(d.spectra['e'].values, 0, 0)

counts = d.spectra[f'n{species}']
```

```python
energies_mid = (energy_bins[:-1] + energy_bins[1:]) / 2
energy_diff = np.diff(energy_bins)

plt.figure()

for ts in range(len(my_steps)):
    plt.loglog(energies_mid, counts.isel(t = ts) / energy_diff, label = f"t = {my_steps[ts]}")

plt.axvline(run.sigma, color = 'black', linestyle = '--')

plt.title(boilerplate + f"\n ylim = {run.ylim}")
plt.xlabel("$\gamma - 1$")
plt.ylabel("$dN/d\gamma$")
plt.legend()
plt.show()
```
